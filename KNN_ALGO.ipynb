#Algorithm - KNN 
#Definition:
Algorithm that classifies data points by comparing them with 'k' nearest neighbors and then does the prediction,uses euclidean/manhatan to find the distance.
#Introduction:
-Supervised ML algo,used for both regression and classification.
-KNN tries to predict the correct class for test data by calculating distance between test data and training points,after that k no. of points get selected which are closest to test data,and at last the knn algo calculates the probability the class of test data belonging to the class of K-training data and the class with high probability get chosen.
#Algorithm:
1.Select the no.K of neighbours.
2.Then calculates euclidean distance of k no. of neighbours,take k nearest neighbours as per euclidean distance.
3.Among these k neigbours,count for no. of data points and then assign the new data points to that category for which the number of the neighbor is maximum.

```python
X = df.drop(columns=["Status"])  
y = df["Status"]                 
```

```python
df
```

```python
from sklearn.preprocessing import StandardScaler, LabelEncoder
for column in X.select_dtypes(include=['object']).columns:
    X['Race'] = LabelEncoder().fit_transform(X['Race'])
```

```python
df
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)
```

```python
y_pred = knn.predict(X_test)
```

```python
print("Accuracy:", accuracy_score(y_test, y_pred))
```

```python
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```
